```python
import pandas as pd
from pathlib import Path

data_path = Path.cwd().parent / "data" / "raw"

print("Pandas version:", pd.__version__)
print("Data path:", data_path)
print("Train file exists:", (data_path / "train.csv").exists())
```

```python
# Reload main dataset
df = pd.read_csv(data_path / "train.csv")

# Convert date
df["date"] = pd.to_datetime(df["date"])

# Create date features
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day
df["day_of_week"] = df["date"].dt.dayofweek

# Reload supporting datasets
stores = pd.read_csv(data_path / "stores.csv")
oil = pd.read_csv(data_path / "oil.csv")
transactions = pd.read_csv(data_path / "transactions.csv")

# Convert dates
oil["date"] = pd.to_datetime(oil["date"])
transactions["date"] = pd.to_datetime(transactions["date"])

# Clean oil data
oil = oil.sort_values("date").reset_index(drop=True)
oil["dcoilwtico"] = oil["dcoilwtico"].interpolate(
    method="linear",
    limit_direction="both"
)

# Merge datasets
df = df.merge(stores, on="store_nbr", how="left")
df = df.merge(oil, on="date", how="left")
df = df.merge(
    transactions,
    on=["date", "store_nbr"],
    how="left"
)

print("Merged Dataset Shape:", df.shape)
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Create complete daily date range
all_dates = pd.date_range(
    start=df["date"].min(),
    end=df["date"].max(),
    freq="D"
)

# Reindex oil dataset to include every date
oil_full = (
    oil.set_index("date")
    .reindex(all_dates)
)

oil_full.index.name = "date"

# Interpolate missing oil prices
oil_full["dcoilwtico"] = oil_full["dcoilwtico"].interpolate(
    method="linear",
    limit_direction="both"
)

# Remove old oil column
df = df.drop(columns=["dcoilwtico"])

# Merge corrected oil dataset
df = df.merge(
    oil_full.reset_index(),
    on="date",
    how="left"
)

# Verify
print("Oil Missing Values:", df["dcoilwtico"].isnull().sum())
print("Dataset Shape:", df.shape)
```

```python
# Analyze missing transactions

missing_transactions = df[df["transactions"].isnull()]

print("Missing Transaction Rows:", len(missing_transactions))

print("\nUnique Missing Dates:",
      missing_transactions["date"].nunique())

print("\nMissing Dates Sample:")
print(
    missing_transactions["date"]
    .drop_duplicates()
    .sort_values()
    .head(15)
)

print("\nMissing Dates by Year:")
print(
    missing_transactions.groupby("year")["date"]
    .nunique()
)

print("\nMissing Transactions by Store:")
print(
    missing_transactions.groupby("store_nbr")
    .size()
    .head(15)
)
```

```python
# Identify missing store-date combinations

expected = df[["date", "store_nbr"]].drop_duplicates()

actual = transactions[["date", "store_nbr"]].drop_duplicates()

comparison = expected.merge(
    actual,
    on=["date", "store_nbr"],
    how="left",
    indicator=True
)

missing_combinations = comparison[
    comparison["_merge"] == "left_only"
]

print("Expected Store-Date Records:", len(expected))
print("Available Transaction Records:", len(actual))
print("Missing Store-Date Records:", len(missing_combinations))

print("\nMissing Records by Store:")
print(
    missing_combinations.groupby("store_nbr")
    .size()
    .sort_values(ascending=False)
)

print("\nMissing Records by Year:")
print(
    missing_combinations.groupby(
        missing_combinations["date"].dt.year
    ).size()
)
```

```python
# Investigate transaction availability by store

selected_stores = [52, 22, 42, 21, 29]

print("Transaction History by Store:\n")

for store_id in selected_stores:
    store_data = transactions[
        transactions["store_nbr"] == store_id
    ]

    print(f"Store {store_id}")
    print("Records:", len(store_data))

    if len(store_data) > 0:
        print("First Date:", store_data["date"].min())
        print("Last Date:", store_data["date"].max())
    else:
        print("No transaction records found")

    print("-" * 40)

# Check store information
display(
    stores[stores["store_nbr"].isin(selected_stores)]
)
```

```python
# Analyze store sales before transaction history starts

selected_stores = [52, 22, 42, 21, 29]

for store_id in selected_stores:

    first_transaction_date = transactions[
        transactions["store_nbr"] == store_id
    ]["date"].min()

    store_sales = df[
        df["store_nbr"] == store_id
    ]

    before_start = store_sales[
        store_sales["date"] < first_transaction_date
    ]

    after_start = store_sales[
        store_sales["date"] >= first_transaction_date
    ]

    print(f"\nStore {store_id}")
    print("First Transaction Date:", first_transaction_date)

    print("Sales Before Transaction History:")
    print("Total Sales:", before_start["sales"].sum())
    print("Rows:", len(before_start))

    print("Sales After Transaction History:")
    print("Total Sales:", after_start["sales"].sum())
    print("Rows:", len(after_start))

    print("-" * 45)
```

```python
# Get first transaction date for each store
first_dates = (
    transactions.groupby("store_nbr")["date"]
    .min()
    .reset_index(name="first_transaction_date")
)

# Add opening information to missing combinations
missing_analysis = missing_combinations.merge(
    first_dates,
    on="store_nbr",
    how="left"
)

# Separate missing records
pre_opening = missing_analysis[
    missing_analysis["date"] < missing_analysis["first_transaction_date"]
]

post_opening = missing_analysis[
    missing_analysis["date"] >= missing_analysis["first_transaction_date"]
]

print("Pre-Opening Missing Records:", len(pre_opening))
print("Post-Opening Missing Records:", len(post_opening))

print("\nPost-Opening Missing Records by Store:")
print(
    post_opening.groupby("store_nbr").size()
    .sort_values(ascending=False)
)

# Check sales on post-opening missing dates
post_sales = post_opening.merge(
    df.groupby(["date", "store_nbr"])["sales"]
    .sum()
    .reset_index(),
    on=["date", "store_nbr"],
    how="left"
)

print("\nSales on Post-Opening Missing Dates:")
print(post_sales["sales"].describe())
```

```python
# Calculate sales for pre-opening missing dates
pre_sales = pre_opening.merge(
    df.groupby(["date", "store_nbr"])["sales"]
    .sum()
    .reset_index(),
    on=["date", "store_nbr"],
    how="left"
)

print("Pre-Opening Records:", len(pre_sales))

print("\nSales Distribution:")
print(pre_sales["sales"].describe())

print("\nNon-Zero Sales Records:",
      (pre_sales["sales"] > 0).sum())

print("\nZero Sales Records:",
      (pre_sales["sales"] == 0).sum())
```

```python
# Fill only validated pre-opening missing transactions

df = df.merge(
    first_dates,
    on="store_nbr",
    how="left"
)

pre_opening_mask = (
    df["transactions"].isnull() &
    (df["date"] < df["first_transaction_date"])
)

df.loc[pre_opening_mask, "transactions"] = 0

# Remove temporary column
df.drop(columns=["first_transaction_date"], inplace=True)

# Verify missing values
print("Remaining Missing Transactions:",
      df["transactions"].isnull().sum())

print("Filled Pre-Opening Records:",
      pre_opening_mask.sum())

print("Final Dataset Shape:", df.shape)
```

```python
# Separate holiday categories
national = holidays[
    (holidays["locale"] == "National") &
    (~holidays["transferred"])
][["date"]].drop_duplicates()

national["national_holiday"] = 1

regional = holidays[
    (holidays["locale"] == "Regional") &
    (~holidays["transferred"])
][["date", "locale_name"]].drop_duplicates()

regional = regional.rename(
    columns={"locale_name": "state"}
)
regional["regional_holiday"] = 1

local = holidays[
    (holidays["locale"] == "Local") &
    (~holidays["transferred"])
][["date", "locale_name"]].drop_duplicates()

local = local.rename(
    columns={"locale_name": "city"}
)
local["local_holiday"] = 1

# Merge holiday features
df_model = df_model.merge(
    national,
    on="date",
    how="left"
)

df_model = df_model.merge(
    regional,
    on=["date", "state"],
    how="left"
)

df_model = df_model.merge(
    local,
    on=["date", "city"],
    how="left"
)

# Fill non-holiday dates with zero
holiday_columns = [
    "national_holiday",
    "regional_holiday",
    "local_holiday"
]

df_model[holiday_columns] = (
    df_model[holiday_columns].fillna(0).astype("int8")
)

# Verify
print("Dataset Shape:", df_model.shape)

print("\nHoliday Feature Counts:")
print(df_model[holiday_columns].sum())

print("\nMissing Values:")
print(df_model[holiday_columns].isnull().sum())
```

```python
# Validate holiday features

holiday_columns = [
    "national_holiday",
    "regional_holiday",
    "local_holiday"
]

# Create combined holiday indicator
df_model["is_holiday"] = (
    df_model[holiday_columns].max(axis=1)
)

print("Unique Holiday Dates:")
print(
    df_model.loc[
        df_model["is_holiday"] == 1, "date"
    ].nunique()
)

print("\nRows with Multiple Holiday Flags:")
print(
    (df_model[holiday_columns].sum(axis=1) > 1).sum()
)

print("\nCombined Holiday Rows:")
print(df_model["is_holiday"].sum())

print("\nHoliday Feature Sample:")
display(
    df_model.loc[
        df_model["is_holiday"] == 1,
        ["date", "city", "state"] + holiday_columns
    ].head(10)
)

print("\nDataset Shape:", df_model.shape)
```

```python
# Advanced calendar features

df_model["quarter"] = df_model["date"].dt.quarter

df_model["week_of_year"] = (
    df_model["date"].dt.isocalendar().week.astype("int16")
)

df_model["day_of_year"] = (
    df_model["date"].dt.dayofyear
)

# Saturday = 5, Sunday = 6
df_model["is_weekend"] = (
    df_model["day_of_week"].isin([5, 6]).astype("int8")
)

# Verify new features
calendar_features = [
    "quarter",
    "week_of_year",
    "day_of_year",
    "is_weekend"
]

print("New Calendar Features:")
display(df_model[calendar_features].head(10))

print("\nWeekend Distribution:")
print(df_model["is_weekend"].value_counts())

print("\nQuarter Distribution:")
print(df_model["quarter"].value_counts().sort_index())

print("\nDataset Shape:", df_model.shape)
```

```python
# Create promotion indicator
df_model["is_promotion"] = (
    df_model["onpromotion"] > 0
).astype("int8")

# Promotion analysis
print("Promotion Distribution:")
print(df_model["is_promotion"].value_counts())

print("\nPromotion Percentage:")
print(
    (df_model["is_promotion"].value_counts(normalize=True) * 100)
    .round(2)
)

print("\nAverage Sales by Promotion:")
print(
    df_model.groupby("is_promotion")["sales"]
    .mean()
)

print("\nAverage Sales by Promotion Quantity:")
print(
    df_model.groupby("is_promotion")["onpromotion"]
    .mean()
)

print("\nDataset Shape:", df_model.shape)
```

```python
# Create lag features

daily_sales = daily_sales.sort_values("date").copy()

daily_sales["sales_lag_1"] = (
    daily_sales["total_sales"].shift(1)
)

daily_sales["sales_lag_7"] = (
    daily_sales["total_sales"].shift(7)
)

daily_sales["sales_lag_28"] = (
    daily_sales["total_sales"].shift(28)
)

print("Updated Dataset Shape:", daily_sales.shape)

display(
    daily_sales[
        ["date", "total_sales",
         "sales_lag_1", "sales_lag_7", "sales_lag_28"]
    ].head(35)
)
```

```python
# Create rolling mean features

daily_sales["rolling_mean_7"] = (
    daily_sales["total_sales"]
    .shift(1)
    .rolling(window=7)
    .mean()
)

daily_sales["rolling_mean_28"] = (
    daily_sales["total_sales"]
    .shift(1)
    .rolling(window=28)
    .mean()
)

print("Updated Dataset Shape:", daily_sales.shape)

display(
    daily_sales[
        [
            "date",
            "total_sales",
            "rolling_mean_7",
            "rolling_mean_28"
        ]
    ].head(35)
)
```

```python
# Create rolling standard deviation features

daily_sales["rolling_std_7"] = (
    daily_sales["total_sales"]
    .shift(1)
    .rolling(window=7)
    .std()
)

daily_sales["rolling_std_28"] = (
    daily_sales["total_sales"]
    .shift(1)
    .rolling(window=28)
    .std()
)

print("Updated Dataset Shape:", daily_sales.shape)

display(
    daily_sales[
        [
            "date",
            "total_sales",
            "rolling_std_7",
            "rolling_std_28"
        ]
    ].head(35)
)
```
